```python
import os
import json
from parser.parser import *
from parser.utils import clean_newlines
from parser.scaffolder import *

def preprocess_markdown(markdown):
    markdown = markdown.replace("*", "").replace("## ", "").replace("“", "\"").replace("”", "\"").strip()
    markdown = clean_newlines(markdown)
    return markdown

def parsing_pipeline(markdown):
    markdown = preprocess_markdown(markdown)
    # pprint(markdown) #DEBUG
    legal_tree = parser(markdown)
    return legal_tree
```

```python
from markitdown import MarkItDown
import io

md = MarkItDown()
def convert2markdown(html_content):
    file_stream = io.BytesIO(html_content.encode("utf-8"))
    markdown = md.convert(file_stream, file_extension=".html")
    return markdown.text_content
```

Tách Điểm + Gộp text Khoản

```python
def collect_all_text(node):
    """Hàm đệ quy gom toàn bộ text của node hiện tại và tất cả các node con."""
    texts = []
    
    # Lấy text của node nếu có
    node_text = node.get("text", "").strip()
    if node_text:
        texts.append(node_text)
    
    # Đệ quy xuống các node con (khoản, điểm,...)
    for child in node.get("children", []):
        child_text = collect_all_text(child)
        if child_text:
            texts.append(child_text)
            
    return "\n".join(texts)


def get_list_dieu(nodes):
    if not nodes:
        return []

    result = []

    for node in nodes:
        if node.get("partType") == "điều":
            result.append({
                "partId": node.get("partId"),
                "partType": node.get("partType"),
                "treeId": node.get("treeId"),
                "text": collect_all_text(node)
            })
            continue
        result.extend(get_list_dieu(node.get("children", [])))

    return result
```

Lọc văn bản có Phụ lục

```python
def clean_appendix_from_list_dieu(list_dieu):
    """
    Hàm duyệt qua list_dieu, giữ lại chuỗi Điều tăng dần (1, 2, 3...).
    Ngay khi partId quay đầu về số nhỏ hơn hoặc bằng số trước đó (ví dụ 1, 2, 1...),
    toàn bộ phần phía sau sẽ bị cắt bỏ vì thuộc Phụ lục/Mục lục dính kèm.
    """
    cleaned_list = []
    last_seen_id = 0

    for dieu in list_dieu:
        part_id_raw = dieu.get("partId")
        
        try:
            current_id = int(part_id_raw)
        except (ValueError, TypeError):
            # Ngay khi gặp partId không phải số (ví dụ chuỗi bất thường), coi như kết thúc phần Điều chính
            break

        # Nếu partId hiện tại lớn hơn partId liền trước -> Hợp lệ (1, 2, 3...)
        if current_id > last_seen_id:
            cleaned_list.append(dieu)
            last_seen_id = current_id
        else:
            # Ngay khi quay về 1 hoặc nhỏ hơn/bằng last_seen_id -> Bắt đầu Phụ lục -> Dừng hẳn
            break

    return cleaned_list
```

```python
modified_docs_count = 0

for doc in tqdm(documents, desc="Cleaning appendix from list_dieu"):
    original_list_dieu = doc.get("list_dieu", [])
    if not original_list_dieu:
        continue

    cleaned_list = clean_appendix_from_list_dieu(original_list_dieu)

    if len(cleaned_list) < len(original_list_dieu):
        modified_docs_count += 1

    doc["list_dieu"] = cleaned_list

print(f"Tổng số văn bản: {len(documents):,}")
print(f"Số văn bản đã loại bỏ thành công phần Phụ lục: {modified_docs_count:,}")
```

```python
from sentence_transformers import SentenceTransformer
import torch

model = SentenceTransformer("AITeamVN/Vietnamese_Embedding_v2", device="cuda" if torch.cuda.is_available() else "cpu")
model.max_seq_length = 2048
print("Device:", model.device)
print("Max seq length:", model.max_seq_length)
```

EDA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm

MAX_SEQ_LENGTH = 2048

tokenizer = model.tokenizer

print("Tokenizer:", tokenizer.name_or_path)
print("Model max length:", tokenizer.model_max_length)


token_lengths = []

for doc in tqdm(documents, desc="Counting tokens"):
    for dieu in (doc["list_dieu"]):
        text = dieu.get("text", "")
        tokens = tokenizer(
            text,
            add_special_tokens=True,
            truncation=False
        )["input_ids"]

        token_lengths.append({
            "doc_id": doc["id"],
            "dieu_id": dieu.get("partId"),
            "token_count": len(tokens),
            "text": text
        })

df_tokens = pd.DataFrame(token_lengths)
```

Chunk điều > 2048 tokens

```python
oversized_pairs = {(check["doc_id"], str(check["dieu_id"]))  for check in tokens_over_2048}

chunked_docs_count = 0

for doc in tqdm(documents, desc="Processing and chunking oversized Dieus"):
    doc_id = doc["id"]
    new_list_dieu = []
    has_chunked = False

    for dieu in doc.get("list_dieu", []):
        dieu_id = str(dieu.get("partId"))
        
        # Kiểm tra xem Điều này có nằm trong danh sách > 2048 tokens không
        if (doc_id, dieu_id) in oversized_pairs:
            # Chỉ chunk Điều bị oversized này
            chunked_khoans = chunker([dieu])
            new_list_dieu.extend(chunked_khoans)
            has_chunked = True
        else:
            # Giữ nguyên Điều hợp lệ
            new_list_dieu.append(dieu)
            
    if has_chunked:
        chunked_docs_count += 1
        
    doc["list_dieu"] = new_list_dieu

print(f"Tổng số văn bản: {len(documents):,}")
print(f"Số văn bản có Điều bị chunk nhỏ theo Khoản: {chunked_docs_count:,}")
```

```python
def get_list_khoan(nodes):
    if not nodes:
        return []

    result = []

    for node in nodes:
        if node.get("partType") == "khoản":
            result.append({
                "partId": node.get("partId"),
                "partType": node.get("partType"),
                "treeId": node.get("treeId"),
                "text": collect_all_text(node)
            })
            continue
        result.extend(get_list_khoan(node.get("children", [])))

    return result
```

```python
from dataclasses import asdict

def chunker(list_dieu):
    list_khoan = []
    
    for dieu in list_dieu:
        raw_text = dieu.get("text", "").strip()
        if not raw_text:
            continue
            
        # 1. Trích xuất dòng tiêu đề của Điều (Dòng đầu tiên, ví dụ: "Điều 1. Phạm vi điều chỉnh")
        dieu_lines = raw_text.split("\n")
        dieu_title = dieu_lines[0].strip()
        
        # 2. Parse lại văn bản Điều thành cấu trúc cây các Khoản
        parse_dieu = parser(raw_text)
        parse_dieu["content"]["parts"] = [asdict(part) for part in parse_dieu["content"]["parts"]]
        
        list_object_khoan = get_list_khoan(parse_dieu["content"]["parts"])
        
        # 3. Nếu Điều có các Khoản con, ghép dieu_title vào từng Khoản
        if list_object_khoan:
            for obj_khoan in list_object_khoan:
                text_khoan = obj_khoan.get("text", "").strip()
                
                # Ghép Tiêu đề Điều + Nội dung Khoản
                chunk_text = f"{dieu_title}\n{text_khoan}"
                
                dieu_chunk = {
                    "partId": dieu["partId"],
                    "partType": dieu["partType"],
                    "treeId": dieu["treeId"],
                    "khoanId": obj_khoan.get("partId"),  # (Tùy chọn) Lưu thêm ID của Khoản
                    "text": chunk_text
                }
                list_khoan.append(dieu_chunk)
        else:
            dieu_chunk = {
                "partId": dieu["partId"],
                "partType": dieu["partType"],
                "treeId": dieu["treeId"],
                "text": raw_text
            }
            list_khoan.append(dieu_chunk)
            
    return list_khoan
```

```python
for document in tqdm(documents):
    list_dieu = document.get("list_dieu", [])
    texts = [f"passage: {dieu['text']}" for dieu in list_dieu]

    if not texts:
        document["list_dieu_vector"] = []
        continue

    vectors = model.encode(
        texts,
        batch_size=8,
        normalize_embeddings=True,
        show_progress_bar=False
    )

    document["list_dieu_vector"] = []

    for dieu, vector in zip(list_dieu, vectors):
        document["list_dieu_vector"].append({
            "partId": dieu["partId"],
            "partType": dieu["partType"],
            "treeId": dieu["treeId"],
            "text": dieu["text"],
            "vector": vector.tolist()
        })
```
